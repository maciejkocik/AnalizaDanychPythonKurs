# <span style="color:#0f766e; font-weight:700;">1. Rozkłady stosowane w statystyce</span>

W statystyce klasycznej kilka rozkładów pojawia się wyjątkowo często — nie dlatego że są matematycznie wygodne, lecz dlatego że wynikają bezpośrednio z danych i procedur wnioskowania.

## <span style="color:#0f766e; font-weight:700;">Przegląd</span>

| Rozkład | Symbol | Zastosowanie |
|---|---|---|
| Normalny | N(μ, σ) | rozkład danych ciągłych, aproksymacja, CTG |
| t-Studenta | t(df) | wnioskowanie o μ przy nieznanym σ, małe próby |
| chi-kwadrat | χ²(df) | testy zgodności, wnioskowanie o wariancji |
| F | F(df₁, df₂) | porównanie wariancji, ANOVA |
| Dwumianowy | Bin(n,p) | dane dychotomiczne, testy proporcji |
| Poissona | Poi(λ) | dane zliczeniowe, rzadkie zdarzenia |

## <span style="color:#0f766e; font-weight:700;">Skąd biorą się t, χ², F?</span>

Wszystkie trzy wywodzą się z rozkładu normalnego:
- **χ²(k)** = suma kwadratów k niezależnych zmiennych N(0,1),
- **t(k)** = N(0,1) / √(χ²(k)/k) — iloraz normalnej i pierwiastka ze znormalizowanej chi-kwadrat,
- **F(k₁,k₂)** = (χ²(k₁)/k₁) / (χ²(k₂)/k₂) — iloraz dwóch znormalizowanych chi-kwadrat.

# <span style="color:#0f766e; font-weight:700;">2. Rozkład t-Studenta</span>

## <span style="color:#0f766e; font-weight:700;">Definicja i parametry</span>

**t(df)** — rozkład t-Studenta z `df` stopniami swobody.
- `stats.t(df=k)`
- E[X] = 0 (dla df > 1), Var[X] = df/(df−2) (dla df > 2)

## <span style="color:#0f766e; font-weight:700;">Kluczowa właściwość: ciężkie ogony</span>

Dla małego df rozkład t ma grubsze ogony niż N(0,1) — zdarzenia ekstremalne są bardziej prawdopodobne.
Wraz ze wzrostem df rozkład t zbiega do N(0,1). Dla df ≥ 30 różnica jest praktycznie pomijalna.

## <span style="color:#0f766e; font-weight:700;">Kiedy używamy?</span>

Gdy szacujemy lub testujemy średnią μ populacji, ale **nie znamy σ** i szacujemy ją z próby:
- statystyka testowa t = (x̄ − μ₀) / (s / √n) ma rozkład t(n−1),
- gdzie s to odchylenie standardowe z próby.

## <span style="color:#0f766e; font-weight:700;">Porównanie ogonów</span>

| df | P(\|t\| > 2) | P(\|t\| > 3) |
|---|---|---|
| 1 | 0.295 | 0.205 |
| 5 | 0.102 | 0.030 |
| 10 | 0.074 | 0.015 |
| 30 | 0.055 | 0.006 |
| ∞ (N(0,1)) | 0.046 | 0.003 |

# <span style="color:#0f766e; font-weight:700;">3. Rozkład chi-kwadrat</span>

## <span style="color:#0f766e; font-weight:700;">Definicja i parametry</span>

**χ²(df)** — suma kwadratów `df` niezależnych zmiennych N(0,1).
- `stats.chi2(df=k)`
- E[X] = df, Var[X] = 2·df
- Rozkład jest **silnie prawostronnie skośny** dla małych df; symetryzuje się przy dużych df.

## <span style="color:#0f766e; font-weight:700;">Kiedy używamy?</span>

- **Test zgodności χ²**: czy dane empiryczne pasują do oczekiwanego rozkładu?
- **Test niezależności**: czy dwie zmienne kategoryczne są niezależne?
- **Wnioskowanie o wariancji**: przedział ufności dla σ².

## <span style="color:#0f766e; font-weight:700;">Kluczowe kwantyle</span>

Dla df=10:
- χ²(0.025) ≈ 3.25 (dolny kraniec przedziału 95%)
- χ²(0.975) ≈ 20.48 (górny kraniec przedziału 95%)

W scipy: `stats.chi2(df=10).ppf(0.025)` i `stats.chi2(df=10).ppf(0.975)`.

# <span style="color:#0f766e; font-weight:700;">4. Symulacje Monte Carlo</span>

## <span style="color:#0f766e; font-weight:700;">Idea</span>

**Metoda Monte Carlo** polega na wielokrotnym losowaniu i zliczaniu zdarzeń, żeby oszacować prawdopodobieństwa lub wartości, których nie potrafimy (lub nie chcemy) obliczyć analitycznie.

Schemat:
1. Zdefiniuj eksperyment losowy.
2. Powtórz go N razy (N = 10 000–1 000 000).
3. Policz, jak często zaszło interesujące zdarzenie.
4. Oszacowanie: P(zdarzenie) ≈ liczba_sukcesów / N.

## <span style="color:#0f766e; font-weight:700;">Zastosowania</span>

- Szacowanie prawdopodobieństwa złożonych zdarzeń,
- Numeryczne całkowanie (gdy całka analityczna jest trudna),
- Symulacja procesów finansowych, fizycznych, biologicznych,
- Weryfikacja wzorów i twierdzeń probabilistycznych.

## <span style="color:#0f766e; font-weight:700;">Dokładność</span>

Błąd standardowy oszacowania MC: SE ≈ √(p(1−p)/N).
Dla N=10 000 i p≈0.5: SE ≈ 0.005 — dokładność do 2 miejsc po przecinku.

# <span style="color:#0f766e; font-weight:700;">5. Centralne Twierdzenie Graniczne</span>

## <span style="color:#0f766e; font-weight:700;">Twierdzenie (wersja intuicyjna)</span>

Jeśli losujemy **niezależne próby** rozmiaru n z **dowolnego** rozkładu o skończonej wariancji, to rozkład **średniej próbkowej** x̄ zbiega do rozkładu normalnego gdy n → ∞:

$$\bar{X} \sim N\!\left(\mu,\, \frac{\sigma}{\sqrt{n}}\right)$$

gdzie μ i σ to parametry rozkładu populacji.

## <span style="color:#0f766e; font-weight:700;">Co to oznacza w praktyce?</span>

- **Nie musimy znać rozkładu populacji** — jeśli n jest wystarczająco duże, średnia i tak ma rozkład normalny.
- Reguła praktyczna: n ≥ 30 zwykle wystarcza (wyjątek: bardzo skośne rozkłady).
- Im bardziej symetryczny rozkład populacji, tym szybsza zbieżność.

## <span style="color:#0f766e; font-weight:700;">Warunki CTG</span>

1. Próby są **niezależne**.
2. Rozkład populacji ma **skończoną wariancję**.
3. n jest **wystarczająco duże** (zależy od skośności rozkładu).

# <span style="color:#0f766e; font-weight:700;">6. Błąd standardowy</span>

## <span style="color:#0f766e; font-weight:700;">Definicja</span>

**Błąd standardowy** (SE, *Standard Error*) to odchylenie standardowe rozkładu próbkowania statystyki.

Dla średniej:

$$SE(\bar{X}) = \frac{\sigma}{\sqrt{n}}$$

Jeśli σ jest nieznane, szacujemy je przez s (odchylenie z próby):

$$\widehat{SE} = \frac{s}{\sqrt{n}}$$

## <span style="color:#0f766e; font-weight:700;">Interpretacja</span>

- SE mierzy, jak bardzo średnia z próby **różni się od próby do próby**.
- Im większa próba, tym mniejszy SE — wyniki są bardziej powtarzalne.
- SE **≠** odchylenie standardowe danych (s mierzy rozproszenie obserwacji, SE mierzy precyzję oszacowania).

## <span style="color:#0f766e; font-weight:700;">Konsekwencje</span>

| Wzrost n (4×) | Efekt na SE |
|---|---|
| n → 4n | SE → SE/2 (dwukrotna poprawa) |
| n → 9n | SE → SE/3 |
| n → 100n | SE → SE/10 |

Poprawienie precyzji 10-krotnie wymaga 100-krotnie większej próby — **prawo malejących zwrotów**.

# <span style="color:#0f766e; font-weight:700;">8. Przykłady</span>

## <span style="color:#0f766e; font-weight:700;">Przykład 1: Rozkład t-Studenta — porównanie z rozkładem normalnym</span>

Cięższe ogony rozkładu t oznaczają, że ekstremalne wartości są bardziej prawdopodobne niż w N(0,1). Efekt jest wyraźny dla małego df i zanika przy df ≥ 30.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

x = np.linspace(-5, 5, 400)
dist_norm = stats.norm(0, 1)

# Wykres porównawczy pdf
plt.figure(figsize=(10, 4))
plt.plot(x, dist_norm.pdf(x), color='black', lw=2.5, linestyle='--', label='N(0,1)', zorder=5)
kolory = ['tomato', 'darkorange', 'steelblue', 'mediumseagreen']
for df, kolor in zip([1, 5, 15, 30], kolory):
    plt.plot(x, stats.t(df=df).pdf(x), color=kolor, lw=1.8, label=f't(df={df})')

plt.xlabel('x')
plt.ylabel('Gęstość')
plt.title('Rozkład t-Studenta vs N(0,1) — porównanie ogonów')
plt.legend()
plt.tight_layout()
plt.show()

# Prawdopodobieństwo ogonów: P(|t| > 2)
print("P(|X| > 2):")
print(f"  N(0,1)  : {2*(1 - dist_norm.cdf(2)):.4f}")
for df in [1, 5, 15, 30]:
    p = 2 * (1 - stats.t(df=df).cdf(2))
    print(f"  t(df={df:2d}): {p:.4f}")
```

## <span style="color:#0f766e; font-weight:700;">Przykład 2: Rozkład chi-kwadrat — pdf i kluczowe kwantyle</span>

Chi-kwadrat jest prawostronnie skośny dla małych df. Kwantyle χ² są używane w przedziałach ufności dla wariancji i testach zgodności.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

x = np.linspace(0, 40, 500)

# Wykres pdf dla różnych df
plt.figure(figsize=(10, 4))
kolory = ['tomato', 'darkorange', 'steelblue', 'mediumseagreen', 'purple']
for df, kolor in zip([1, 3, 5, 10, 20], kolory):
    dist = stats.chi2(df=df)
    y = dist.pdf(x)
    plt.plot(x, y, color=kolor, lw=1.8, label=f'χ²(df={df})')

plt.xlabel('x')
plt.ylabel('Gęstość')
plt.title('Rozkład chi-kwadrat dla różnych stopni swobody')
plt.legend()
plt.xlim(0, 40)
plt.ylim(0, 0.5)
plt.tight_layout()
plt.show()

# Kluczowe kwantyle dla df=10
df = 10
dist = stats.chi2(df=df)
print(f"\nKwantyle χ²(df={df}):")
for q in [0.025, 0.05, 0.50, 0.95, 0.975]:
    print(f"  ppf({q:.3f}) = {dist.ppf(q):.4f}")

print(f"\nE[X]   = {dist.mean():.2f}  (= df = {df})")
print(f"Var[X] = {dist.var():.2f}  (= 2·df = {2*df})")
```

## <span style="color:#0f766e; font-weight:700;">Przykład 3: Monte Carlo — estymacja liczby π</span>

Rzucamy punkty losowo w kwadrat [−1,1]². Punkt trafia w koło jeśli x²+y²≤1. Stosunek pól daje π/4.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
N = 50_000

# Losowanie punktów w kwadracie [-1, 1]²
x = rng.uniform(-1, 1, size=N)
y = rng.uniform(-1, 1, size=N)

# Punkt jest w kole, jeśli x² + y² ≤ 1
w_kole = (x**2 + y**2) <= 1.0

# Estymacja: π ≈ 4 * (liczba punktów w kole) / N
pi_est = 4 * w_kole.mean()
print(f"N = {N:,}")
print(f"Szacowane π = {pi_est:.5f}")
print(f"Prawdziwe  π = {np.pi:.5f}")
print(f"Błąd absolutny = {abs(pi_est - np.pi):.5f}")

# Wizualizacja (tylko 3000 punktów dla czytelności)
n_vis = 3000
fig, ax = plt.subplots(figsize=(6, 6))
ax.scatter(x[:n_vis][~w_kole[:n_vis]], y[:n_vis][~w_kole[:n_vis]],
           s=0.8, color='tomato', alpha=0.5, label='poza kołem')
ax.scatter(x[:n_vis][w_kole[:n_vis]],  y[:n_vis][w_kole[:n_vis]],
           s=0.8, color='steelblue', alpha=0.5, label='w kole')
theta = np.linspace(0, 2*np.pi, 300)
ax.plot(np.cos(theta), np.sin(theta), 'k-', lw=1.5)
ax.set_aspect('equal')
ax.set_title(f'Monte Carlo: π ≈ {pi_est:.4f}  (N={N:,})')
ax.legend(markerscale=8)
plt.tight_layout()
plt.show()
```

## <span style="color:#0f766e; font-weight:700;">Przykład 4: CTG — demonstracja dla rozkładu jednostajnego</span>

Rozkład jednostajny jest płaski i bardzo odległy od normalnego. Pokażemy, że rozkład średnich próbkowych szybko zbiega do normalnego.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

rng = np.random.default_rng(42)
n_symulacji = 10_000

# Parametry rozkładu jednostajnego Uniform(0, 1)
mu_pop = 0.5        # E[X] = (a+b)/2
sigma_pop = 1/np.sqrt(12)  # SD = (b-a)/sqrt(12)

fig, axes = plt.subplots(1, 4, figsize=(14, 4))
n_values = [1, 5, 30, 100]

for ax, n in zip(axes, n_values):
    # Losujemy n_symulacji próbek rozmiaru n, obliczamy średnią każdej
    probki = rng.uniform(0, 1, size=(n_symulacji, n))
    srednie = probki.mean(axis=1)

    # Histogram rozkładu średnich
    sns.histplot(srednie, stat='density', bins=40, ax=ax, color='steelblue', alpha=0.5)

    # Nakładamy teoretyczną aproksymację normalną (CTG)
    se = sigma_pop / np.sqrt(n)
    x_plot = np.linspace(srednie.min(), srednie.max(), 200)
    ax.plot(x_plot, stats.norm(mu_pop, se).pdf(x_plot), color='tomato', lw=2)
    ax.set_title(f'n={n}')
    ax.set_xlabel('Średnia próbki')
    ax.set_ylabel('Gęstość')

fig.suptitle('CTG: rozkład średnich z Uniform(0,1) — zbieżność do normalnego', y=1.02)
plt.tight_layout()
plt.show()
```

## <span style="color:#0f766e; font-weight:700;">Przykład 5: CTG — demonstracja dla rozkładu Poissona (skośnego)</span>

Rozkład Poissona jest prawostronnie skośny. CTG działa wolniej niż dla rozkładu jednostajnego.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

rng = np.random.default_rng(42)
n_symulacji = 10_000
lam = 2.0  # parametr Poissona — skośny rozkład

mu_pop = lam
sigma_pop = np.sqrt(lam)

fig, axes = plt.subplots(1, 4, figsize=(14, 4))

for ax, n in zip(axes, [1, 5, 30, 100]):
    # Losowanie próbek rozmiaru n z Poissona
    probki = rng.poisson(lam, size=(n_symulacji, n))
    srednie = probki.mean(axis=1)

    sns.histplot(srednie, stat='density', bins=40, ax=ax, color='mediumseagreen', alpha=0.5)
    se = sigma_pop / np.sqrt(n)
    x_plot = np.linspace(max(0, srednie.min()), srednie.max(), 200)
    ax.plot(x_plot, stats.norm(mu_pop, se).pdf(x_plot), color='tomato', lw=2)
    ax.set_title(f'n={n}')
    ax.set_xlabel('Średnia próbki')
    ax.set_ylabel('Gęstość')

fig.suptitle(f'CTG: rozkład średnich z Poisson(λ={lam}) — skośny rozkład populacji', y=1.02)
plt.tight_layout()
plt.show()
```

## <span style="color:#0f766e; font-weight:700;">Przykład 6: Błąd standardowy — jak maleje z n</span>

SE = σ/√n. Pokażemy, że SE maleje proporcjonalnie do 1/√n — poprawienie precyzji wymaga coraz większych nakładów.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

sigma = 15.0  # odchylenie standardowe populacji (np. wyniki IQ)
n_values = np.arange(1, 501)

# SE = sigma / sqrt(n)
se_values = sigma / np.sqrt(n_values)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Wykres SE(n) — skala liniowa
axes[0].plot(n_values, se_values, color='steelblue', lw=2)
axes[0].axhline(sigma/np.sqrt(30), color='tomato', linestyle='--', lw=1.5, label=f'SE(n=30) = {sigma/np.sqrt(30):.2f}')
axes[0].axhline(sigma/np.sqrt(100), color='darkorange', linestyle='--', lw=1.5, label=f'SE(n=100) = {sigma/np.sqrt(100):.2f}')
axes[0].set_xlabel('Rozmiar próby (n)')
axes[0].set_ylabel('Błąd standardowy (SE)')
axes[0].set_title('SE = σ/√n  (σ=15)')
axes[0].legend()

# Wykres SE(n) — skala log-log (powinna być linią prostą)
axes[1].loglog(n_values, se_values, color='steelblue', lw=2)
axes[1].set_xlabel('n (skala log)')
axes[1].set_ylabel('SE (skala log)')
axes[1].set_title('SE(n) w skali log-log')

plt.tight_layout()
plt.show()

# Tabela: ile razy musisz zwiększyć n, żeby zmniejszyć SE 2x
print(f"SE przy n=25:  {sigma/np.sqrt(25):.3f}")
print(f"SE przy n=100: {sigma/np.sqrt(100):.3f}  (4× więcej danych → 2× mniejszy SE)")
print(f"SE przy n=400: {sigma/np.sqrt(400):.3f}  (16× więcej danych → 4× mniejszy SE)")
```

## <span style="color:#0f766e; font-weight:700;">Przykład 7: Monte Carlo — rozkład sumy trzech kostek</span>

Analityczne obliczenie rozkładu sumy 3 kostek jest żmudne. Monte Carlo daje wynik w kilku linijkach.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
N = 200_000

# Rzut trzema kostkami sześciennymi, N razy
kostki = rng.integers(1, 7, size=(N, 3))
sumy = kostki.sum(axis=1)  # suma każdego rzutu

# Rozkład empiryczny
unikalne, liczby = np.unique(sumy, return_counts=True)
czestosci = liczby / N

print(f"P(suma >= 15) = {(sumy >= 15).mean():.4f}")
print(f"P(suma == 3)  = {(sumy == 3).mean():.5f}  (tylko 1,1,1)")
print(f"E[suma]       = {sumy.mean():.3f}  (teoretycznie: 3 * 3.5 = 10.5)")

# Wykres
plt.figure(figsize=(10, 4))
plt.bar(unikalne, czestosci, color='steelblue', alpha=0.8)
plt.xlabel('Suma trzech kostek')
plt.ylabel('Częstość empiryczna')
plt.title(f'Rozkład sumy 3 kostek sześciennych (N={N:,})')
plt.xticks(unikalne)
plt.tight_layout()
plt.show()
```

```python
# 1. Definiujemy rozkłady i ich parametry
dfs = [3, 10, 30]
punkty_odciecia = [1.96, 2.576]

dane_tabeli = []

# Obliczenia dla rozkładów t-Studenta
for df in dfs:
    # Dla testu dwustronnego: P(|X| > c) = 2 * (1 - cdf(c))
    p_196 = 2 * (1 - stats.t.cdf(1.96, df=df))
    p_2576 = 2 * (1 - stats.t.cdf(2.576, df=df))
    
    dane_tabeli.append({
        "df": f"t(df={df})",
        "P(|X| > 1.96)": f"{p_196:.4f}",
        "P(|X| > 2.576)": f"{p_2576:.4f}"
    })

# Obliczenia dla rozkładu normalnego N(0,1)
p_norm_196 = 2 * (1 - stats.norm.cdf(1.96))
p_norm_2576 = 2 * (1 - stats.norm.cdf(2.576))

dane_tabeli.append({
    "df": "N(0,1)",
    "P(|X| > 1.96)": f"{p_norm_196:.4f}",
    "P(|X| > 2.576)": f"{p_norm_2576:.4f}"
})

# 2. Tworzenie i wyświetlenie tabeli danych
df_tabela = pd.DataFrame(dane_tabeli)
print(df_tabela.to_string(index=False))

```

### Komentarz:
- Różnice w prawdopodobieństwach maleją wraz ze wzrostem liczby stopni swobody (df), ponieważ rozkład t-Studenta jest asymptotycznie zbieżny do rozkładu normalnego N(0,1).
- Przy małym df rozkład t ma tzw. "grubsze ogony" (większa niepewność wynikająca z małej próby).
- Im większa próba (wyższe df), tym dokładniej szacujemy odchylenie standardowe populacji, przez co rozkład t-Studenta zaczyna idealnie pokrywać się z rozkładem standardowym normalnym.

Instrukcja:

   - P(−t* < X < t*) = 0.90 (przedział 90%),
   - P(−t* < X < t*) = 0.95 (przedział 95%),
   - P(−t* < X < t*) = 0.99 (przedział 99%).
2. Powtórz to samo dla N(0,1).
3. Wydrukuj wyniki w postaci tabeli. Dla każdego wiersza zaznacz, o ile % wartość t jest większa od z.

Wymagane funkcje:
- `stats.t(df=15).ppf(q)` — dla przedziału symetrycznego P(−t*<X<t*)=0.95 użyj q=0.975
- `stats.norm(0,1).ppf(q)`

Kryterium zaliczenia:
- poprawne kwantyle dla t(15) i N(0,1),
- kolumna z procentową różnicą.

Checkpoint:
- pokaż tabelę i wskaż, przy której ufności różnica jest największa.

Dokumentacja:
- https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.t.html

```python
# 1. Definiujemy parametry
df_value = 15
poziomy_ufnosci = [0.90, 0.95, 0.99]
tabela_wynikow = []

# 2. Obliczamy kwantyle dla każdego poziomu ufności
for conf in poziomy_ufnosci:
    # Dla przedziału symetrycznego: q = 1 - (1 - poziom_ufnosci) / 2
    q = 1 - (1 - conf) / 2
    
    t_val = stats.t(df=df_value).ppf(q)
    z_val = stats.norm(0, 1).ppf(q)
    
    # Procentowa różnica: o ile procent t* jest większe od z*
    roznica_procentowa = ((t_val - z_val) / z_val) * 100
    
    tabela_wynikow.append({
        "Poziom ufności": f"{conf*100:.0f}%",
        "Kwantyl t* (df=15)": round(t_val, 4),
        "Kwantyl z* (N(0,1))": round(z_val, 4),
        "Różnica (%)": f"+{roznica_procentowa:.2f}%"
    })

# 3. Tworzenie i wyświetlenie tabeli
df_tabela = pd.DataFrame(tabela_wynikow)
print(df_tabela.to_string(index=False))
```

Różnica procentowa między kwantylem rozkładu t-Studenta a rozkładu normalnego jest największa dla poziomu ufności $99\%$ i wynosi aż $+14.40\%$ (wartość kwantyla $t^* = 2.9467$ w porównaniu do $z^* = 2.5758$).

Dlaczego tak się dzieje? Wynika to bezpośrednio z cechy charakterystycznej rozkładu t-Studenta, jaką są tzw. „grube ogony” (heavy tails). 

Im dalej przesuwamy się w stronę krańców rozkładu (zwiększając poziom ufności z $90\%$ do $99\%$), tym wolniej gęstość prawdopodobieństwa rozkładu t-Studenta spada w porównaniu do rozkładu normalnego $N(0,1)$.

W rezultacie, aby odciąć skrajne $1\%$ prawdopodobieństwa (dla poziomu ufności $99\%$), musimy odsunąć się na osi poziomej znacznie dalej w rozkładzie t-Studenta niż w standardowym rozkładzie normalnym. Różnica ta staje się tym bardziej widoczna, im wyższy poziom ufności analizujemy.

## <span style="color:#0f766e; font-weight:700;">Zadanie 4</span>

Monte Carlo — estymacja π metodą Buffona.

Instrukcja:
1. Zasymuluj estymację π metodą Buffona (igła Buffona):
   - Rzucamy igłę długości L=1 na podłogę z równoległymi liniami co d=2 (L < d).
   - Środek igły: y ~ Uniform(0, d/2), kąt: θ ~ Uniform(0, π/2).
   - Igła przecina linię, jeśli: y ≤ (L/2)·sin(θ).
   - P(przecięcie) = 2L / (π·d), stąd π ≈ 2L·N / (d·liczba_przecięć).
2. Zasymuluj N=100 000 rzutów.
3. Wydrukuj oszacowanie π i błąd absolutny.
4. Narysuj zbieżność oszacowania (kumulowaną) dla kolejnych rzutów.

Wymagane funkcje:
- `rng.uniform()`
- `np.cumsum()`
- `plt.plot()` ze skalą log na osi x

Kryterium zaliczenia:
- poprawna formuła estymacji π,
- wykres zbieżności kumulowanej.

Checkpoint:
- pokaż wykres i oszacowanie π prowadzącemu.

Dokumentacja:
- https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.uniform.html

```python
rng = np.random.default_rng(seed=42)

N = 100000
L = 1.0  # długość igły
d = 2.0  # odległość między liniami

# Środek igły: y ~ Uniform(0, d/2)
y = rng.uniform(0, d/2, size=N)

# Kąt: theta ~ Uniform(0, pi/2)
theta = rng.uniform(0, np.pi/2, size=N)

# Igła przecina linię, jeśli y <= (L/2) * sin(theta)
przeciecia = y <= (L/2) * np.sin(theta)

# Obliczenie zbieżności kumulowanej dla kolejnych rzutów
skumulowane_przeciecia = np.cumsum(przeciecia)
numery_rzutow = np.arange(1, N + 1)

# Wzór: pi ≈ (2 * L * n) / (d * liczba_przecięć)
zbieznosc_pi = np.zeros(N)
maska = skumulowane_przeciecia > 0
zbieznosc_pi[maska] = (2 * L * numery_rzutow[maska]) / (d * skumulowane_przeciecia[maska])

# Wyniki końcowe
koncowe_pi = zbieznosc_pi[-1]
blad_absolutny = abs(koncowe_pi - np.pi)

print(f"WYNIKI SYMULACJI (N = {N:,}):")
print(f"Oszacowana wartość π: {koncowe_pi:.6f}")
print(f"Prawdziwa wartość π: {np.pi:.6f}")
print(f"Błąd absolutny: {blad_absolutny:.6f}")

# Generowanie wykresu zbieżności kumulowanej
plt.figure(figsize=(10, 5))
plt.plot(numery_rzutow, zbieznosc_pi, color='#0f766e', linewidth=1.5, label='Estymacja π (Buffon)')
plt.axhline(y=np.pi, color='#dc2626', linestyle='--', linewidth=1.5, label='Teoretyczna wartość π')

plt.xscale('log')

plt.title('Zbieżność estymacji π metodą igły Buffona', fontsize=13, fontweight='bold')
plt.xlabel('Liczba rzutów (skala logarytmiczna)', fontsize=11)
plt.ylabel('Wartość π', fontsize=11)
plt.ylim(2.5, 3.8) 
plt.grid(True, which="both", linestyle='--', alpha=0.5)
plt.legend(fontsize=10)

plt.show()
```

Monte Carlo — symulacja złożonego zdarzenia.

Instrukcja:
1. W urnie jest 10 kul: 4 czerwone, 3 niebieskie, 3 zielone.
2. Losujemy 4 kule **bez zwracania**.
3. Za pomocą symulacji Monte Carlo (N=100 000) oszacuj:

   - P(wszystkie 4 tego samego koloru),
   - P(dokładnie 1 zielona i dokładnie 1 niebieska).
4. Porównaj z wynikiem analitycznym dla pierwszego prawdopodobieństwa (kombinatoryka).

Wymagane funkcje:

- `np.sum()`, `np.mean()`

Kryterium zaliczenia:
- trzy poprawne oszacowania MC,
- wynik analityczny P(≥2 czerwone) obliczony i porównany.

```python
import scipy.special as special

# Definicja urny i parametrów symulacji
urna = np.array([0]*4 + [1]*3 + [2]*3)
N = 100000

# Używamy Generatora do losowania bez zwracania dla N prób jednocześnie
rng = np.random.default_rng(seed=42)

# Wektorowe losowanie bez zwracania: generujemy losowe permutacje indeksów i bierzemy pierwsze 4
macierz_losowa = rng.random((N, 10))
indeksy_losowania = np.argsort(macierz_losowa, axis=1)[:, :4]
wyniki_symulacji = urna[indeksy_losowania]

# Zliczanie liczby poszczególnych kolorów w każdym z N losowań
liczba_czerwonych = np.sum(wyniki_symulacji == 0, axis=1)
liczba_niebieskich = np.sum(wyniki_symulacji == 1, axis=1)
liczba_zielonych   = np.sum(wyniki_symulacji == 2, axis=1)

# Obliczanie prawdopodobieństw z symulacji Monte Carlo
p_przynajmniej_2_czerwone_mc = np.mean(liczba_czerwonych >= 2)
p_wszystkie_ten_sam_kolor_mc = np.mean((liczba_czerwonych == 4) | (liczba_niebieskich == 4) | (liczba_zielonych == 4))
p_dokladnie_1z_i_1n_mc       = np.mean((liczba_zielonych == 1) & (liczba_niebieskich == 1))

# Wszystkie możliwe sposoby wyciągnięcia 4 kul z 10: Kombinacja(10, 4)
wszystkie_mozliwosci = special.comb(10, 4)

# Zdarzenie przeciwne: 0 czerwonych lub 1 czerwona
sposoby_0_czerwonych = special.comb(4, 0) * special.comb(6, 4)
sposoby_1_czerwona   = special.comb(4, 1) * special.comb(6, 3)

# Sposoby na co najmniej 2 czerwone
sukcesy_analityczne = wszystkie_mozliwosci - sposoby_0_czerwonych - sposoby_1_czerwona
p_przynajmniej_2_czerwone_analit = sukcesy_analityczne / wszystkie_mozliwosci

# 4. Wydruk wyników i porównanie
print("WYNIKI SYMULACJI MONTE CARLO (N = 100 000):")
print(f"P(co najmniej 2 czerwone) = {p_przynajmniej_2_czerwone_mc:.5f}")
print(f"P(wszystkie 4 tego samego koloru) = {p_wszystkie_ten_sam_kolor_mc:.5f}")
print(f"P(dokładnie 1 zielona i dokładnie 1 niebieska) = {p_dokladnie_1z_i_1n_mc:.5f}")
print("\n")
print("PORÓWNANIE Z WYNIKIEM ANALITYCZNYM:")
print(f"Wartość analityczna P(>=2 czerwone) = {p_przynajmniej_2_czerwone_analit:.5f}")
print(f"Różnica bezwzględna (Błąd MC) = {abs(p_przynajmniej_2_czerwone_mc - p_przynajmniej_2_czerwone_analit):.5f}")
```

## <span style="color:#0f766e; font-weight:700;">Zadanie 6</span>

CTG — demonstracja dla rozkładu chi-kwadrat.

Instrukcja:
1. Użyj rozkładu chi-kwadrat χ²(df=2) jako populacji (silnie skośny, E[X]=2, Var[X]=4).
2. Dla n ∈ {5, 15, 50, 200} zasymuluj 8000 próbek rozmiaru n.
3. Oblicz średnią każdej próbki.
4. Narysuj 4 histogramy w siatce 2×2. Na każdym nałóż teoretyczne pdf aproksymacji normalnej.
5. Dodaj podtytuł figury z nazwą rozkładu populacji.

Wymagane funkcje:
- `stats.chi2(df=2).rvs(size=(8000, n), random_state=rng)`
- `.mean(axis=1)`
- `sns.histplot(stat='density')`

Kryterium zaliczenia:
- 4 histogramy w siatce 2×2,
- widoczna zbieżność do normalnego.

Checkpoint:
- pokaż wykresy i wskaż, dla którego n aproksymacja jest już dobra.

Dokumentacja:
- https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.chi2.html

```python
rng = np.random.default_rng(seed=42)

# Parametry populacji chi-kwadrat (df=2)
df_pop = 2
mu = 2 # E[X] = df = 2
wariancja = 4   # Var[X] = 2 * df = 4
sigma = np.sqrt(wariancja)  # sigma = 2

# Rozmiary prób n do przetestowania
n_sizes = [5, 15, 50, 200]

# Tworzenie siatki wykresów 2x2
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()  

for i, n in enumerate(n_sizes):
    # Generowanie 8000 próbek rozmiaru n
    probki = stats.chi2(df=df_pop).rvs(size=(8000, n), random_state=rng)
    
    # Obliczanie średniej dla każdej z 8000 próbek (wzdłuż wierszy)
    srednie_probkowe = probki.mean(axis=1)
    
    # Rysowanie histogramu gęstości
    sns.histplot(srednie_probkowe, stat='density', ax=axes[i], color='#0f766e', alpha=0.6, bins=40, edgecolor='white')
    
    # Wyznaczenie parametrów teoretycznych aproksymacji normalnej: N(mu, sigma / sqrt(n))
    se = sigma / np.sqrt(n)
    x_teor = np.linspace(srednie_probkowe.min(), srednie_probkowe.max(), 500)
    pdf_teor = stats.norm(loc=mu, scale=se).pdf(x_teor)
    
    # Nałożenie krzywej teoretycznej rozkładu normalnego
    axes[i].plot(x_teor, pdf_teor, color='#dc2626', linewidth=2.5, label='Teoretyczne PDF')
    
    # Formatowanie pojedynczego podwykresu
    axes[i].set_title(f'Rozmiar próby n = {n}', fontsize=12, fontweight='bold')
    axes[i].set_xlabel('Średnia z próby')
    axes[i].set_ylabel('Gęstość')
    axes[i].grid(True, linestyle='--', alpha=0.5)
    if i == 0:
        axes[i].legend()

fig.suptitle('CTG dla populacji o rozkładzie chi-kwadrat (df=2)', fontsize=14, fontweight='bold', y=0.96)

# Korekta układu przed zapisem
plt.tight_layout(rect=[0, 0.03, 1, 0.93])
plt.show()
```

### Dla którego n aproksymacja jest już dobra?
- Dla n=5 oraz n=15: Widoczna jest wyraźna asymetria (prawostronna skośność) histogramu. Czerwona krzywa teoretycznego rozkładu normalnego (która z natury jest symetryczna) nie nadąża za lewą stroną rozkładu i "ucieka" w prawo na szczycie. Oznacza to, że dla małych prób rozkład średniej wciąż pamięta asymetryczny kształt populacji pierwotnej.
- Dla n=50: Histogram zaczyna układać się w bardzo ładny, symetryczny dzwon. Odchylenia od teoretycznej krzywej normalnej są już minimalne.
- Dla n=200: Aproksymacja jest bardzo dobra (niemal idealna). Histogram perfekcyjnie pokrywa się z wykresem gęstości rozkładu normalnego.

Wniosek: Bezpieczną granicą, przy której możemy uznać aproksymację za w pełni satysfakcjonującą dla tak skośnego rozkładu, jest wartość w okolicach n=50.

### Dlaczego zmniejszenie SE 2× wymaga 4× więcej danych?

Błąd standardowy średniej ($SE$) maleje nieliniowo wraz ze wzrostem wielkości próby ($n$). Ponieważ parametr $n$ znajduje się pod pierwiastkiem kwadratowym w mianowniku wzoru ($SE = \frac{\sigma}{\sqrt{n}}$), obowiązuje tzw. prawo pierwiastka kwadratowego. Oznacza to, że dwukrotne zmniejszenie błędu (zwiększenie precyzji) wymaga aż czterokrotnego zwiększenia liczby zebranych danych. Zjawisko to ilustruje punkt $n=100$ ($SE=2.0$), gdzie dalsze powiększanie próby przynosi coraz mniejsze korzyści przy gwałtownie rosnących kosztach pozyskania danych.

## <span style="color:#0f766e; font-weight:700;">Zadanie 10</span>

Monte Carlo — całkowanie numeryczne.

Instrukcja:
1. Oblicz całkę ∫₀¹ f(x) dx, gdzie f(x) = 4√(1−x²) (ćwiartka koła jednostkowego — wynik = π).
2. Metoda MC:
   - losuj x ~ Uniform(0,1),
   - oblicz f(x) = 4*sqrt(1−x²) dla każdego x,
   - oszacowanie: ∫f(x)dx ≈ mean(f(x)).
3. Uruchom dla N ∈ {100, 1000, 10000, 100000}. Dla każdego wydrukuj oszacowanie i błąd.
4. Narysuj zbieżność (błąd absolutny vs N) na skali log-log.

Wymagane funkcje:
- `rng.uniform(0, 1, size=N)`
- `np.sqrt()`, `np.mean()`
- `plt.loglog()`

Kryterium zaliczenia:
- poprawne oszacowania dla 4 wartości N,
- wykres log-log zbieżności błędu.

Dokumentacja:
- https://numpy.org/doc/stable/reference/generated/numpy.mean.html

```python
# 1. Inicjalizacja stabilnego generatora liczb pseudolosowych
rng = np.random.default_rng(seed=42)

# Wartości liczby próbek N oraz dokładna wartość całki (pi)
rozmiary_N = [100, 1000, 10000, 100000]
prawdziwa_wartosc = np.pi

oszacowania = []
bledy_absolutne = []

print("WYNIKI CAŁKOWANIA METODĄ MONTE CARLO:")

# 2. Pętla obliczeniowa dla różnych wartości N
for N in rozmiary_N:
    # Losowanie x z rozkładu jednostajnego Uniform(0, 1)
    x = rng.uniform(0, 1, size=N)
    
    # Obliczanie wartości funkcji f(x) = 4 * sqrt(1 - x^2)
    f_x = 4 * np.sqrt(1 - x**2)
    
    # Oszacowanie całki jako średnia z wartości funkcji
    oszacowanie = np.mean(f_x)
    blad = abs(oszacowanie - prawdziwa_wartosc)
    
    oszacowania.append(oszacowanie)
    bledy_absolutne.append(blad)
    
    print(f"N = {N:6d} | Oszacowanie = {oszacowanie:.6f} | Błąd = {blad:.6f}")


# 3. Rysowanie wykresu zbieżności błędu w skali log-log
plt.loglog(rozmiary_N, bledy_absolutne, marker='o', color='#0f766e', linewidth=2, label='Błąd symulacji')

# Opcjonalnie: nałożenie teoretycznej linii zbieżności 1/sqrt(N) dla celów porównawczych
linia_teoretyczna = 1 / np.sqrt(rozmiary_N)
# Skalujemy linię teoretyczną, aby pasowała wizualnie do wykresu bledu
linia_teoretyczna = linia_teoretyczna * (bledy_absolutne[0] / linia_teoretyczna[0])
plt.loglog(rozmiary_N, linia_teoretyczna, linestyle='--', color='#dc2626', label='Teoretyczny trend 1/sqrt(N)')

# Tytuły i opisy osi
plt.title('Zbieżność błędu całkowania numerycznego Monte Carlo', fontsize=12, fontweight='bold')
plt.xlabel('Liczba prób (N) - skala logarytmiczna', fontsize=11)
plt.ylabel('Błąd absolutny - skala logarytmiczna', fontsize=11)
plt.grid(True, which="both", linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

## <span style="color:#0f766e; font-weight:700;">Zadanie 11</span>

Rozkład t-Studenta w praktyce — przedział ufności dla średniej.

Wprowadzenie:
Gdy szacujemy μ z małej próby i nie znamy σ, używamy rozkładu t. Przedział ufności (1−α) dla μ:
[x̄ − t* · s/√n,  x̄ + t* · s/√n]
gdzie t* = ppf(1−α/2) dla t(df=n−1).

Instrukcja:
1. Wygeneruj próbę n=15 z N(50, 8) (udajemy, że nie znamy σ).
2. Oblicz x̄ i s z próby.
3. Wyznacz t* dla ufności 95% (df=n−1=14).
4. Oblicz i wydrukuj przedział ufności 95%.
5. Powtórz 1000 razy; policz, w jakim % przypadków prawdziwa μ=50 jest wewnątrz przedziału.

Wymagane funkcje:
- `stats.t(df=n-1).ppf(0.975)`
- `np.mean()`, `np.std(ddof=1)`

Kryterium zaliczenia:
- poprawny przedział ufności dla jednej próby,
- procent pokrycia bliski 95% dla 1000 powtórzeń.

Checkpoint:
- pokaż przedział i procent pokrycia.

Dokumentacja:
- https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.t.html

```python
# 1. Definiowanie stałych parametrów populacji
mu_prawdziwa = 50
sigma_prawdziwa = 8
n = 15
poziom_ufnosci = 0.95
alpha = 1 - poziom_ufnosci

# Zainicjowanie stabilnego generatora liczb pseudolosowych
rng = np.random.default_rng(seed=42)

# Generowanie próby n=15 z rozkładu normalnego
proba = rng.normal(mu_prawdziwa, sigma_prawdziwa, size=n)

# Obliczanie średniej (x_bar) oraz odchylenia standardowego z próby (s) z ddof=1
x_bar = np.mean(proba)
s = np.std(proba, ddof=1)

# Wyznaczanie wartości krytycznej t* dla df = n - 1 = 14
t_gwiazdka = stats.t(df=n-1).ppf(1 - alpha/2)

# Obliczanie błędu maksymalnego oraz granic przedziału ufności
blad_maksymalny = t_gwiazdka * (s / np.sqrt(n))
dolna_granica = x_bar - blad_maksymalny
gorna_granica = x_bar + blad_maksymalny

print("WYNIKI DLA POJEDYNCZEJ PRÓBY (n = 15):")
print(f"Średnia z próby (x_bar): {x_bar:.4f}")
print(f"Odchylenie standardowe z próby (s): {s:.4f}")
print(f"Wartość krytyczna t* (df=14): {t_gwiazdka:.4f}")
print(f"Przedział ufności 95%: [{dolna_granica:.4f}, {gorna_granica:.4f}]")
print(f"Czy prawdziwa średnia (50) wewnątrz?: {dolna_granica <= mu_prawdziwa <= gorna_granica}")

# Symulacja Monte Carlo (1000 powtórzeń)

licznik_pokrycia = 0
liczba_powtorzen = 1000

for _ in range(liczba_powtorzen):
    # Generowanie nowej, niezależnej próbki za każdym razem
    sim_proba = rng.normal(mu_prawdziwa, sigma_prawdziwa, size=n)
    sim_x_bar = np.mean(sim_sample := sim_proba)
    sim_s = np.std(sim_proba, ddof=1)
    
    # Wyznaczenie przedziału dla bieżącej próbki
    sim_blad = t_gwiazdka * (sim_s / np.sqrt(n))
    sim_dolna = sim_x_bar - sim_blad
    sim_gorna = sim_x_bar + sim_blad
    
    # Sprawdzenie, czy prawdziwa wartość mu znajduje się w przedziale
    if sim_dolna <= mu_prawdziwa <= sim_gorna:
        licznik_pokrycia += 1

procent_pokrycia = (licznik_pokrycia / liczba_powtorzen) * 100

print("\nWYNIKI SYMULACJI MONTE CARLO (1000 POWTÓRZEŃ):")
print(f"Liczba przedziałów zawierających mu=50: {licznik_pokrycia} / {liczba_powtorzen}")
print(f"Empiryczny procent pokrycia: {procent_pokrycia:.2f}%")
print(f"Oczekiwany procent pokrycia (teoria): {poziom_ufnosci * 100:.2f}%")
```

## <span style="color:#0f766e; font-weight:700;">Zadanie 12</span>

Monte Carlo — problem Sekretarki (optymalne zatrzymanie).

Wprowadzenie:
Przeglądam kolejno n=20 kandydatów. Widzę ich jakość (1–100), ale decyzja jest nieodwracalna. Strategia: odrzuć pierwszych k kandydatów (faza obserwacji), potem wybierz pierwszego lepszego od najlepszego w fazie obserwacji.

Instrukcja:
1. Zasymuluj strategię dla k ∈ {0, 1, 2, …, 19} (faza odrzucenia = k kandydatów).
2. Dla każdego k oblicz (N=50 000 symulacji) prawdopodobieństwo wyboru najlepszego kandydata.
3. Narysuj wykres P(wybór najlepszego) jako funkcję k.
4. Znajdź optymalne k i porównaj z teorią: k* ≈ n/e ≈ 7.

Wymagane funkcje:
- `rng.permutation(n)` — losowa kolejność kandydatów
- `np.max()` — najlepszy z fazy obserwacji

Kryterium zaliczenia:
- wykres prawdopodobieństwa vs k,
- zidentyfikowane optymalne k.

Checkpoint:
- pokaż wykres; opisz, gdzie jest maksimum i jak blisko jest n/e.

Dokumentacja:
- https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.permutation.html

```python
# 1. Inicjalizacja stabilnego generatora liczb pseudolosowych
rng = np.random.default_rng(seed=42)
n = 20       
N = 50000  

# Generowanie N niezależnych losowych permutacji kandydatów.

permutacje = np.array([rng.permutation(n) for _ in range(N)])

prawdopodobienstwa_sukcesu = []

# 2. Ewaluacja strategii dla każdej możliwej wielkości fazy obserwacji k
for k in range(n):
    if k == 0:
        # Brak fazy obserwacji: bezwarunkowo akceptujemy pierwszego kandydata z brzegu
        wybrani_kandydaci = permutacje[:, 0]
    else:
        # Wyznaczenie najlepszego kandydata w fazie obserwacji (pierwsze k kolumn)
        najlepszy_z_obserwacji = np.max(permutacje[:, :k], axis=1)
        
        # Sprawdzamy, którzy kandydaci w fazie właściwej (od indeksu k do końca) są lepsi
        jest_lepszy = permutacje[:, k:] > najlepszy_z_obserwacji[:, None]
        
        # Weryfikacja, czy w ogóle w danej symulacji pojawił się ktoś lepszy
        znaleziono_lepszego = np.any(jest_lepszy, axis=1)
        
        # Znajdujemy indeks pierwszego takiego kandydata
        indeks_pierwszego_lepszego = np.argmax(jest_lepszy, axis=1)
        
        # Domyślnie ustawiamy -1 (brak wyboru / porażka)
        wybrani_kandydaci = np.full(N, -1)
        
        # Dla symulacji, w których pojawił się lepszy kandydat, zapisujemy jego wartość
        wiersze_z_sukcesem = np.arange(N)[znaleziono_lepszego]
        kolumny_z_sukcesem = k + indeks_pierwszego_lepszego[znaleziono_lepszego]
        wybrani_kandydaci[znaleziono_lepszego] = permutacje[wiersze_z_sukcesem, kolumny_z_sukcesem]
        
    # Sukces oznacza wybranie kandydata o najwyższej możliwej wartości (n - 1, czyli 19)
    liczba_sukcesow = np.sum(wybrani_kandydaci == (n - 1))
    prawdopodobienstwa_sukcesu.append(liczba_sukcesow / N)

# Wyznaczenie optymalnego k na podstawie uzyskanych danych z symulacji
optymalne_k = np.argmax(prawdopodobienstwa_sukcesu)
teoretyczne_k = round(n / np.e)

print("WYNIKI SYMULACJI PROBLEMU SEKRETARKI:")
for k, p in enumerate(prawdopodobienstwa_sukcesu):
    oznaczenie = " <- MAKSIMUM" if k == optymalne_k else ""
    print(f"k = {k:2d} | Prawdopodobienstwo sukcesu = {p:.4f}{oznaczenie}")
print(f"Optymalne k z symulacji: {optymalne_k}")
print(f"Teoretyczne k (n/e): {n/np.e:.2f} -> po zaokragleniu: {teoretyczne_k}")

# 3. Rysowanie wykresu prawdopodobieństwa sukcesu w zależności od progu k
plt.plot(range(n), prawdopodobienstwa_sukcesu, marker='o', color='#0f766e', linewidth=2, label='Wyniki MC')

# Zaznaczenie punktu optymalnego na wykresie
plt.axvline(x=optymalne_k, color='#dc2626', linestyle='--', alpha=0.7, 
            label=f'Optymalne k = {optymalne_k}')
plt.plot(optymalne_k, prawdopodobienstwa_sukcesu[optymalne_k], marker='X', color='#dc2626', markersize=10)

plt.title('Problem Sekretarki - Szukanie optymalnego punktu zatrzymania', fontsize=12, fontweight='bold')
plt.xlabel('Liczba odrzuconych kandydatów w fazie obserwacji (k)', fontsize=11)
plt.ylabel('Prawdopodobieństwo wyboru najlepszego kandydata', fontsize=11)
plt.xticks(range(n))
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```

   - oblicz szerokość przedziału opartego na z = 1.96 × SE,
   - zakładaj σ=1 (znana), SE=1/√n.
2. Wydrukuj tabelę: n, szerokość(t), szerokość(z), stosunek t/z.
3. Narysuj obie szerokości jako funkcję n.

Wymagane funkcje:
- `stats.t(df=n-1).ppf(0.975)`
- `stats.norm(0,1).ppf(0.975)` → 1.96

Kryterium zaliczenia:
- poprawna tabela i wykres,
- wniosek: przy jakim n różnica t/z jest < 5%?

Checkpoint:
- pokaż tabelę i wykres; wskaż, od którego n można używać z zamiast t.

Dokumentacja:
- https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.t.html

```python
# 1. Definiowanie parametrów zadania
n_values = [5, 10, 20, 50, 100]
z_star = stats.norm(0, 1).ppf(0.975) 

szerokosci_t = []
szerokosci_z = []
stosunki = []

# Obliczenia dla poszczególnych rozmiarów próby n
for n in n_values:
    se = 1 / np.sqrt(n)  # Błąd standardowy dla założonej sigma=1
    t_star = stats.t(df=n-1).ppf(0.975)  # Kwantyl rozkładu t-Studenta
    
    # Całkowita szerokość przedziału to 2 * margines błędu
    w_t = 2 * t_star * se
    w_z = 2 * z_star * se
    stosunek = w_t / w_z
    
    szerokosci_t.append(w_t)
    szerokosci_z.append(w_z)
    stosunki.append(stosunek)

# 2. Tworzenie i wyświetlenie tabeli wyników

tabela_porownawcza = pd.DataFrame({
    "n": n_values,
    "Szerokość (t)": [round(w, 4) for w in szerokosci_t],
    "Szerokość (z)": [round(w, 4) for w in szerokosci_z],
    "Stosunek t/z": [round(r, 4) for r in stosunki]
})

print("TABELA PORÓWNAWCZA SZEROKOŚCI PRZEDZIAŁÓW:")
print(tabela_porownawcza.to_string(index=False))

n_linia = np.arange(5, 101)
se_linia = 1 / np.sqrt(n_linia)
w_t_linia = 2 * stats.t(df=n_linia-1).ppf(0.975) * se_linia
w_z_linia = 2 * z_star * se_linia

# 3. Rysowanie wykresu funkcji ciągłej
plt.plot(n_linia, w_t_linia, color='#dc2626', linewidth=2, label='Przedział t-Studenta')
plt.plot(n_linia, w_z_linia, color='#0f766e', linewidth=2, linestyle='--', label='Przedział normalny (z)')

plt.scatter(n_values, szerokosci_t, color='#dc2626', zorder=5)
plt.scatter(n_values, szerokosci_z, color='#0f766e', zorder=5)

se_27 = 1 / np.sqrt(27)
w_t_27 = 2 * stats.t(df=26).ppf(0.975) * se_27

# Pionowa linia pomocnicza
plt.axvline(x=27, color='orange', linestyle=':', linewidth=2, label='Granica różnicy 5% (n = 27)')
plt.scatter([27], [w_t_27], color='orange', marker='o', s=80, zorder=6)

# Dodanie strzałki i tekstu objaśniającego za pomocą plt.annotate
plt.annotate('Tutaj różnica spada poniżej 5% (n = 27)', 
             xy=(27, w_t_27), 
             xytext=(42, 1.1),
             arrowprops=dict(facecolor='orange', shrink=0.1, width=1.5, headwidth=7),
             fontsize=10, 
             fontweight='bold')


plt.title('Wpływ wyboru rozkładu na szerokość przedziału ufności 95%', fontsize=11, fontweight='bold')
plt.xlabel('Rozmiar próby (n)', fontsize=10)
plt.ylabel('Szerokość przedziału ufności', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(0, 105)
plt.ylim(0, 2.7)
plt.legend()
```

### Wniosek:

Analizując wyniki z tabeli, widzimy, że dla małej próby $n=20$ szerokość przedziału t-Studenta jest jeszcze o 6.8% większa od przedziału normalnego (stosunek 1.0679). Z kolei dla większej próby $n=50$ różnica ta gwałtownie maleje i wynosi już tylko 2.5% (stosunek 1.
0253). Oznacza to, że punkt, w którym różnica między rozkładami spada poniżej oczekiwanego progu 5%, znajduje się dokładnie pomiędzy tymi dwoma krokami badania. Zaznaczony na wykresie punkt n = 27 stanowi precyzyjną, matematyczną granicę tego przejścia. Od tego miejsca rozkład normalny daje niemal identyczne wyniki co rozkład t-Studenta.

- mylenie SE (błąd standardowy średniej) z s (odchylenie standardowe danych) — SE = s/√n, a nie s,
- użycie N(0,1) zamiast t(df=n−1) przy małej próbie i nieznanym σ — zawyża precyzję przedziałów,
- interpretowanie CTG jako twierdzenia o danych: CTG dotyczy rozkładu **średniej**, nie indywidualnych obserwacji,
- brak `ddof=1` w `np.std()` przy obliczaniu odchylenia z próby — daje obciążony estymator σ,
- zbyt mała liczba symulacji Monte Carlo (N < 1000) — duże błędy losowe, mylące wyniki,
- wnioskowanie z jednej symulacji bez uśredniania — wynik może być daleki od wartości oczekiwanej.
